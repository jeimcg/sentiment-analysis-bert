```python
from textblob import TextBlob
import pandas as pd

# Load the file
file_path = "/Users/jei/sentiment-analysis-ml-excel/data/lyrics_combined.xlsx"
df = pd.read_excel(file_path, sheet_name="Sheet1")

# Split lyrics by line breaks in cell (0,0)
lyrics = df.iloc[0, 0].split('\n')  # Split lyrics into separate lines

# Clean the lyrics text
def clean_text(text):
    import re
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)  # Remove special characters
    return text

lyrics = [clean_text(lyric) for lyric in lyrics if lyric.strip()]  # Remove empty lines

# Add sentiment labels
def get_sentiment(text):
    sentiment = TextBlob(text).sentiment.polarity
    if sentiment > 0:
        return 1  # Positive
    elif sentiment < 0:
        return 0  # Negative
    else:
        return 0.5  # Neutral

# Create a DataFrame with lyrics and labels
labeled_data = pd.DataFrame({
    "Lyric": lyrics,
    "Label": [get_sentiment(lyric) for lyric in lyrics]
})

# Save the labeled dataset to a new Excel file
output_path = "lyrics_with_labels.xlsx"
labeled_data.to_excel(output_path, index=False)

print(f"Labeled dataset saved to {output_path}")
```
